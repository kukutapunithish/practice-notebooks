```python
import nltk
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
doc = nlp(u'Tesla is looking at buying U.S. startup for $6 million')
```

```python
for token in doc:
    print(token.text,token.pos_,token.dep_)
```

```python
nlp.pipeline
```

```python
nlp.pipe_names
```

```python
doc2 = nlp(u"Tesla isn't looking for the startups anymore.")
```

```python
for token in doc2:
    print(token.text,token.pos_,token.dep_,token.shape_,token.is_alpha)
```
