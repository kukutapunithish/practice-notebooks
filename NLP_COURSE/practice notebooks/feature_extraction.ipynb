```python
with open('../03-Text-Classification/1.txt') as f:
    words1 = f.read().split()
with open('../03-Text-Classification/2.txt') as f:
    words2 = f.read().split()
```

```python
vocab = {}
def vocab_build(x,vocab,i):
    for word in x:
        if word in vocab:
            continue
        else:
            vocab[word] = i
            i+=1
    return vocab,i
```

```python
i = 1
vocab,i = vocab_build(words1,vocab,i)
```

```python
vocab,i = vocab_build(words2,vocab,i)
```
