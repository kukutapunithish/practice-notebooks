```python
emails = ['test@example.com',
'alice.bob@example.com',
'user123@domain-name.co.uk',
'first-last@sub.domain.net',
'__test__@example.com']
```

```python
# email = emails[0]
pattern = r'[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+'
for email in emails:
    print(re.fullmatch(pattern,email).group(0))
```

```python
valid_urls = [
    "https://www.geeksforgeeks.org",
    "http://example.com",
    "https://sub.domain.com",
    "ftp://user:password@://ftp.example.com",
    "http://192.168.1.1",
    "https://site.co.uk",
    "https://site.com#section-anchor",
    "https://site.com",
    "https://site.com",
    "http://www.site.museum",
    "http://site.info"
]

invalid_strings = [
    "This is just some random text without a URL.",
    "www.invalid-url.com",
    "http://.com",
    "http://site",
    "https://site.c",
    "just_an_email@example.com",
    "//relative/path",
    "http://user:pass@invalid^domain.com",
    "file://localhost/path/to/file",
    "https://site.com",
    "https://"
]
```

```python
pattern = r"((http|https)://)(www)?[a-zA-Z0-9@:%._\+~#?&//=]{2,256}\.[a-z]{2,6}\b([-a-zA-Z0-9@:%._\+~#?&//=]*)"
# Compile the ReGex
p = re.compile(pattern)

for url in valid_urls:
    print(re.search(p, url))
```
