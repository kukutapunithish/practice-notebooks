```python
df = df.dropna()
```

```python
X = pd.get_dummies(df.drop('species',axis=1))
Y = df['species']
```

```python
X.shape
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,Y,test_size=0.30,random_state=101)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model = DecisionTreeClassifier()
```

```python
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import classification_report,confusion_matrix,ConfusionMatrixDisplay

confusion_matrix(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

```python
ConfusionMatrixDisplay.from_estimator(model,X_test,y_test)
```

```python
model.feature_importances_
```

```python
df_mat = pd.DataFrame(index=X.columns,data=model.feature_importances_,columns=['feature_importance'])
```

```python
df_mat.sort_values(by='feature_importance')
```

```python
from sklearn.tree import plot_tree
```

```python
X_test.head()
```

```python
plt.figure(figsize=(12,8),dpi=200)
plot_tree(model,feature_names=X.columns,filled=True);
```

```python
model = DecisionTreeClassifier(max_depth=3,splitter='best',min_impurity_decrease=0.2)
model.fit(X_train,y_train)
y_pred = model.predict(X_test)
print(classification_report(y_test,y_pred))
plt.figure(figsize=(12,8),dpi=200)
plot_tree(model,feature_names=X.columns,filled=True);
```

```python
model = DecisionTreeClassifier(max_depth=3,splitter='random')
model.fit(X_train,y_train)
y_pred = model.predict(X_test)
print(classification_report(y_test,y_pred))
plt.figure(figsize=(12,8),dpi=200)
plot_tree(model,feature_names=X.columns,filled=True);
```
