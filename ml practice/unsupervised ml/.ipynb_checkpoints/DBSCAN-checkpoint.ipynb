```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
blobs = pd.read_csv('../../Excercises/DATA/cluster_blobs.csv')
```

```python
blobs
```

```python
sns.scatterplot(data=blobs,x='X1',y='X2')
```

```python
moons = pd.read_csv('../../Excercises/DATA/cluster_moons.csv')
```

```python
sns.scatterplot(data=moons,x='X1',y='X2')
```

```python
circles = pd.read_csv('../../Excercises/DATA/cluster_circles.csv')
```

```python
sns.scatterplot(data=circles,x='X1',y='X2')
```

```python
def display_categories(model,data):
    labels = model.fit_predict(data)
    sns.scatterplot(data=data,x='X1',y='X2',hue=labels,palette='Set1')
```

```python
from sklearn.cluster import KMeans

model = KMeans(n_clusters=3)
```

```python
display_categories(model,blobs)
```

```python
model = KMeans(n_clusters=2)
display_categories(model,moons)
```

```python
model = KMeans(n_clusters=2)
display_categories(model,circles)
```

```python
from sklearn.cluster import DBSCAN

model = DBSCAN(eps=0.5,min_samples=2)
```

```python
model = DBSCAN()
display_categories(model,blobs)
```

```python
model = DBSCAN(eps=0.15)
display_categories(model,moons)
```

```python
model = DBSCAN(eps=0.15)
display_categories(model,circles)
```
