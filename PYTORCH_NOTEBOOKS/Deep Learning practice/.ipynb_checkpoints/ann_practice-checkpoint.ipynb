```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
z = np.linspace(-5,5,40)
```

```python
sigmoid_values = 1/(1 + 2.7182**-z)
```

```python
plt.plot(z,sigmoid_values)
```

```python
e = 2.7182
```

```python
cosh = (e**z + e**-z)/2
sinh = (e**z - e**-z)/2
tanh = sinh/cosh
```

```python
plt.plot(z,tanh)
```

```python
def relu(v):
    if v <= 0:
        v = 0.0
    return v
```

```python
# relu_values = z * (z > 0)
# relu_values = np.maximum(0, z)
relu_values = np.vectorize(relu)(z)
```

```python
plt.plot(z,relu_values)
```

```python
prob_vals = []

for zi in z:
    den = 0
    for zj in z:
        den += e**zj
    prob_vals.append(e**zi/den)
```

```python
np.sum(prob_vals)
```
